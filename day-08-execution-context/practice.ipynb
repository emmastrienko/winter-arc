### Exercise 1: Private Counter with Closure Encapsulation

Implement `createCounter(initialValue)` returning an object with `increment()`, `decrement()`, and `getValue()` without exposing `count` directly.


```typescript
export function createCounter(initialValue: number = 0) {
  // TODO: Implement closure-based counter
  let count = initialValue;
  return {
    increment: () => ++count,
    decrement: () => --count,
    getValue: () => count
  };
}
```
