### Exercise 1: Immutable Vector2D with Dunder Protocol

Implement a 2D Vector class supporting `__repr__`, `__str__`, `__eq__`, `__hash__`, `__abs__`, and scalar/vector `__add__`.


```python
import math

class Vector2D:
    def __init__(self, x: float, y: float):
        # TODO: Implement immutable vector storage
        self.x = x
        self.y = y

    def __repr__(self) -> str:
        # TODO: Return 'Vector2D(x=..., y=...)'
        return f"Vector2D(x={float(self.x)}, y={float(self.y)})"

    def __str__(self) -> str:
        # TODO: Return '(x, y)'
        return f"({float(self.x)}, {float(self.y)})"

    def __eq__(self, other: object) -> bool:
        # TODO: Check equality with another Vector2D
        if not isinstance(other, Vector2D):
            return NotImplemented
        return self.x == other.x and self.y == other.y

    def __hash__(self) -> int:
        # TODO: Hash based on x and y
        return hash((self.x, self.y))

    def __abs__(self) -> float:
        # TODO: Return Euclidean magnitude sqrt(x^2 + y^2)
        return math.sqrt(self.x ** 2 + self.y ** 2)

    def __add__(self, other: "Vector2D") -> "Vector2D":
        # TODO: Return new Vector2D representing sum
        if not isinstance(other, Vector2D):
            return NotImplemented
        return Vector2D(self.x + other.x, self.y + other.y)
```

### Exercise 2: Custom Data Sequence: BoundedHistory

Implement a custom sequence `BoundedHistory` with fixed capacity that stores recent items, supporting `__len__`, `__getitem__`, and membership testing.


```python
class BoundedHistory:
    def __init__(self, capacity: int):
        # TODO: Initialize with positive capacity
        if capacity <= 0:
            raise ValueError("Capacity must be a positive integer.")
        self.capacity = capacity
        self.history = []

    def append(self, item: object) -> None:
        # TODO: Add item; if exceeding capacity, drop oldest
        self.history.append(item)
        if len(self.history) > self.capacity:
            self.history.pop(0)

    def __len__(self) -> int:
        return len(self.history)

    def __getitem__(self, index: int) -> object:
        return self.history[index]
```

```python
# Verification cell
history = BoundedHistory(3)
history.append(1)
history.append(2)
history.append(3)
history.append(4)
assert len(history) == 3
assert history[0] == 2
assert history[-1] == 4
assert 3 in history
assert 1 not in history
print("✅ Exercise 2 passed successfully!")
```
