### Exercise 1: Execution Timer Context

Implement `Stopwatch` context manager measuring elapsed seconds.


```python
class Stopwatch:
    def __init__(self):
        self.start_time = None
        self.end_time = None
        self.elapsed_seconds = None
        
    def __enter__(self):
        self.start_time = time.perf_counter()
        self.end_time = None
        self.elapsed_seconds = None
        return self

    def __exit__(self, *args):
        self.end_time = time.perf_counter()
        self.elapsed_seconds = self.end_time - self.start_time
        return False  # Don't suppress exceptions
```

```python
# Verification cell
import time
with Stopwatch() as sw:
    time.sleep(0.02)
assert sw.elapsed_seconds >= 0.01
print("✅ Exercise 1 passed successfully!")
```
