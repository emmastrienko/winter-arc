### Exercise 1: Result Monad Container

Implement `Result[T, E]` with `ok`, `err`, and `unwrap`.


```python
import typing
from typing import Generic, TypeVar

T = TypeVar("T")
E = TypeVar("E")

class Result(Generic[T, E]):
    def __init__(self, value: T | None = None, error: E | None = None, is_ok: bool = True):
        self.value = value
        self.error = error
        self.is_ok = is_ok

    @classmethod
    def ok(cls, value: T) -> "Result[T, E]":
        return cls(value=value, is_ok=True)

    @classmethod
    def error(cls, error: E) -> "Result[T, E]":
        return cls(error=error, is_ok=False)

    @property
    def is_error(self) -> bool:
        return not self.is_ok

    def unwrap(self) -> T:
        if self.is_ok:
            return self.value  # type: ignore
        else:
            raise Exception(f"Called unwrap on an error result: {self.error}")

    def unwrap_error(self) -> E:
        if self.is_error:
            return self.error  # type: ignore
        else:
            raise Exception(f"Called unwrap_error on an ok result: {self.value}")
```

```python
# Verification cell
assert Result.ok(10).unwrap() == 10
print("✅ Exercise 1 passed successfully!")
```
