### Exercise 1: Batch Chunker

Implement `chunked(iterable, size)` yielding lists of size `size`.


```python
def chunked(iterable, size: int):
    if size <= 0:
        raise ValueError("size must be positive")
    
    batch = []
    for item in iterable:
        batch.append(item)
        if len(batch) == size:
            yield batch
            batch = []
    if batch:
        yield batch
```

```python
# Verification cell
assert list(chunked(range(5), 2)) == [[0, 1], [2, 3], [4]]
print("✅ Exercise 1 passed successfully!")
```
