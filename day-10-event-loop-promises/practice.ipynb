### Exercise 1: Promise.all Polyfill

Implement `myPromiseAll(promises)` returning a promise that resolves with an array of values or rejects immediately upon any rejection.


```typescript
export function myPromiseAll<T>(promises: Array<T | Promise<T>>): Promise<T[]> {
  // TODO: Implement myPromiseAll
  return new Promise((resolve, reject) => {
    const results: T[] = [];
    let completedCount = 0;

    if (promises.length === 0) {
      resolve(results);
      return;
    }

    promises.forEach((promise, index) => {
      Promise.resolve(promise)
        .then((value) => {
          results[index] = value;
          completedCount++;
          if (completedCount === promises.length) {
            resolve(results);
          }
        })
        .catch((error) => {
          reject(error);
        });
    });
  });
}
```

```typescript
# Verification cell
const res = await myPromiseAll([Promise.resolve(1), 2, Promise.resolve(3)]);
assert(res[0] === 1 && res[1] === 2 && res[2] === 3);
print("✅ Exercise 1 passed successfully!")
```
