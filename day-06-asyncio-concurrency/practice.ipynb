### Exercise 1: Semaphore-Bounded Batch Runner

Implement `bounded_gather` with `asyncio.Semaphore`.


```python
import asyncio
async def bounded_gather(factories, max_concurrency):
    if max_concurrency <= 0:
        raise ValueError("max_concurrency must be greater than 0")

    sem = asyncio.Semaphore(max_concurrency)

    async def run_one(factory):
        async with sem:
            return await factory()

    return await asyncio.gather(*(run_one(factory) for factory in factories))
```

```python
import asyncio

async def task(x):
    return x * 2

factories = [lambda v=i: task(v) for i in range(3)]

res = await bounded_gather(factories, 2)
assert res == [0, 2, 4]
print("✅ Exercise 1 passed successfully!")
```
